```python
%matplotlib inline 
import numpy as np
from matplotlib_inline import backend_inline
from d2l import torch as d2l
```

```python
def f(x):
    return 3*x**2-4*x #y=3x^2-4x 
```

```python
for h in 10.0**np.arange(-1,-6,-1):   #start,stop,step h ile 
    print(f'h={h:.5f}, numerical limit={(f(1+h)-f(1))/h:.5f}')
```

```python
def usedispl():
    backend_inline.set_matplotlib_formats('svg')
```

```python
def setfigs(figsize=(3.5 , 2.5)): 
    usedispl()
    d2l.plt.rcParams['figure.figsize'] = figsize
```

```python
def setaxes(axes,xlabel,ylabel,xlim,ylim,xscale,yscale,legend):
    axes.set_xlabel(xlabel) , axes.set_ylabel(ylabel)
    axes.set_xscale(xscale) , axes.set_yscale(yscale)
    axes.set_xlim(xlim) , axes.set_ylim(ylim)
    if legend:
        axes.legend(legend)
    axes.grid()
```

```python
def plot (X , Y=None ,xlabel=None, ylabel=None , legend=[],xlim=None,
           ylim=None,xscale='linear',yscale='linear',
           fmts=('-','m--','g-','r:') , figsize=(3.5,2.5),axes=None):
    def hasOneAxis(X):
        return(hasattr(X,"ndim") and X.ndim ==1 or  isinstance(X, list)
               and not hasattr(X[0],"__len__"))
    if hasOneAxis(X): X =[X]
    if Y is None:
        X , Y =[[]]*len(X) , X
    elif hasOneAxis(Y):
        Y=[Y]
    if len(X) !=len(Y):
        X= X*len(Y)
    setfigs(figsize)
    if axes is None:
         axes=d2l.plt.gca()
    axes.cla()
    for x , y, fmt in zip(X,Y,fmts):
        axes.plot(x,y,fmt) if len(x) else axes.plot(y,fmt)
    setaxes(axes, xlabel, ylabel, xlim, ylim, xscale, yscale, legend)


        
    
```

```python
x=np.arange(10, 1334, 23)
plot(x,[f(x) , 2*x-3], 'x','f(x)',legend=['f(x)','Tangent line (x=1)'])
```
